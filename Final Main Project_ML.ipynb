```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.metrics import r2_score, accuracy_score
print("Libraries Imported Successfully!")
data = pd.read_excel("Alternative_Student_Performance_Dataset.xlsx")
print("\nSTUDENT PERFORMANCE DATASET")
print(data)
print("\nDATASET INFORMATION")
data.info()
print("\nMISSING VALUES")
print(data.isnull().sum())
print("\nSTATISTICAL SUMMARY")
print(data.describe())
X = data[['Study Hours', 'Attendance (%)', 'Previous Marks']]
y_reg = data['Final Marks']
y_class = data['Result']
print("\nInput Features:")
print(X.head())
X_train, X_test, y_train, y_test = train_test_split(
    X, y_reg, test_size=0.2, random_state=42
)
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X, y_class, test_size=0.2,
    random_state=42, stratify=y_class
)
print("\nTraining Records:", len(X_train))
print("Testing Records:", len(X_test))
reg_model = LinearRegression()
reg_model.fit(X_train, y_train)
print("\nRegression Model Trained Successfully!")
y_pred = reg_model.predict(X_test)
print("\nACTUAL FINAL MARKS")
print(list(y_test))
print("\nPREDICTED FINAL MARKS")
print(np.round(y_pred, 2))
print("\nREGRESSION MODEL EVALUATION")
print("Mean Absolute Error:",
      round(mean_absolute_error(y_test, y_pred), 2))
print("Mean Squared Error:",
      round(mean_squared_error(y_test, y_pred), 2))
print("R2 Score:",
      round(r2_score(y_test, y_pred), 2))
class_model = LogisticRegression(max_iter=1000)
class_model.fit(X_train_c, y_train_c)
print("\nClassification Model Trained Successfully!")
class_pred = class_model.predict(X_test_c)
print("\nACTUAL RESULTS")
print(list(y_test_c))
print("\nPREDICTED RESULTS")
print(list(class_pred))
print("\nCLASSIFICATION MODEL EVALUATION")
print("Accuracy:",
      round(accuracy_score(y_test_c, class_pred) * 100, 2), "%")
print("\n====================================")
print("LIVE STUDENT PERFORMANCE PREDICTION")
print("====================================")
study = float(input("\nEnter Study Hours: "))
attendance = float(input("Enter Attendance Percentage: "))
previous = float(input("Enter Previous Marks: "))
new_student = pd.DataFrame({
    'Study Hours': [study],
    'Attendance (%)': [attendance],
    'Previous Marks': [previous]
})
print("\nNEW STUDENT DETAILS")
print(new_student)
predicted_marks = reg_model.predict(new_student)
print("\nPredicted Final Marks:",
      round(predicted_marks[0], 2))
predicted_result = class_model.predict(new_student)
print("Predicted Result:", predicted_result[0])
print("\n====================================")
print("STUDENT PERFORMANCE PREDICTION RESULT")
print("====================================")
print("Study Hours:", study)
print("Attendance:", attendance, "%")
print("Previous Marks:", previous)
print("Predicted Final Marks:",
      round(predicted_marks[0], 2))
print("Predicted Pass/Fail:",
      predicted_result[0])
print("====================================")
print("PROJECT COMPLETED SUCCESSFULLY!")
```
